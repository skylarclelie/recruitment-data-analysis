```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
# matplotlib.use('TkAgg')
```

```python
df = pd.read_excel("data/BOSS_Zhipin_Sample_Data.xlsx")
```

```python
# 多少行多少列
print("数据shape：", df.shape)
print()

print("列名：", df.columns.tolist())
print()
print(df.head(3))
```

```python
# 筛选相关岗位
keywords = ['数据', 'Python', 'python', '大数据', '数据分析', '数据开发']

# 只要职位名称里包含这些关键词的
mask = df['职位名称'].str.contains('|'.join(keywords), na=False)
df_target = df[mask]

print("相关岗位总数：", len(df_target))
print()
print("职位名称前20个：")
print(df_target['职位名称'].value_counts().head(20))
```

```python
# 过滤掉标注类岗位
exclude_keywords = ['标注', '训练师', '文员', '采集']

mask2 = ~df_target['职位名称'].str.contains('|'.join(exclude_keywords), na=False)
df_tech = df_target[mask2]

print("过滤后相关岗位数：", len(df_tech))
print()
print("职位名称分布：")
print(df_tech['职位名称'].value_counts().head(20))

print("城市分布：")
print(df_tech['发布城市'].value_counts().head(10))

print("学历要求分布：")
print(df_tech['学历要求'].value_counts())
print()
print("薪资分布：")
print(df_tech['薪资'].value_counts().head(15))
```

```python
# 解决中文显示问题
matplotlib.rcParams['font.sans-serif'] = ['SimHei']
matplotlib.rcParams['axes.unicode_minus'] = False
```

```python
# 学历要求饼图
edu_counts = df_tech['学历要求'].value_counts()

plt.figure(figsize=(8, 6))
plt.pie(edu_counts, labels=edu_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('数据类岗位学历要求分布')
plt.tight_layout()
plt.savefig('学历分布.png')
plt.show()
print("图已保存")
```

```python
# 薪资分布柱状图
salary_counts = df_tech['薪资'].value_counts().head(12)

plt.figure(figsize=(10, 6))
plt.bar(salary_counts.index, salary_counts.values, color='steelblue')
plt.title('数据类岗位薪资分布')
plt.xlabel('薪资范围')
plt.ylabel('岗位数量')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('薪资分布.png')
# plt.show()
print("图已保存")
```

```python
# 经验要求分布
exp_counts = df_tech['经验要求'].value_counts()

plt.figure(figsize=(10, 6))
plt.bar(exp_counts.index, exp_counts.values, color='coral')
plt.title('数据类岗位经验要求分布')
plt.xlabel('经验要求')
plt.ylabel('岗位数量')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('经验分布.png')
print("图已保存")
```

```python
# 把薪资变成数字
# 先看看薪资长什么样
print(df_tech['薪资'].head(10))
```

```python
import re

def parse_salary(s):
    # 去掉·13薪这种后缀
    s = s.split('·')[0]
    # 提取数字
    nums = re.findall(r'\d+', s)
    if len(nums) == 2:
        return (int(nums[0]) + int(nums[1])) / 2
    return None

df_tech = df_tech.copy() 
df_tech['薪资_均值K'] = df_tech['薪资'].apply(parse_salary)

print(df_tech[['职位名称', '薪资', '薪资_均值K']].head(10))
print()
print("平均薪资：", df_tech['薪资_均值K'].mean().round(1), "K")
print("最低中位薪资：", df_tech['薪资_均值K'].min(), "K")
print("最高中位薪资：", df_tech['薪资_均值K'].max(), "K")
# 过滤掉明显异常的薪资（超过200K的不合理）
df_tech_clean = df_tech[df_tech['薪资_均值K'] < 200].copy()

print("过滤后数量：", len(df_tech_clean))
print("平均薪资：", df_tech_clean['薪资_均值K'].mean().round(1), "K")
print("最低薪资：", df_tech_clean['薪资_均值K'].min(), "K")
print("最高薪资：", df_tech_clean['薪资_均值K'].max(), "K")
```

```python
# 薪资箱线图
plt.figure(figsize=(8, 6))
plt.boxplot(df_tech_clean['薪资_均值K'].dropna(), 
            tick_labels=['数据类岗位'],   # 改这里
            patch_artist=True,
            boxprops=dict(facecolor='steelblue', alpha=0.7))
plt.title('数据类岗位薪资分布（K）')
plt.ylabel('薪资（K/月）')
plt.tight_layout()
plt.savefig('薪资箱线图.png')
# plt.show()
print("图已保存")
print()
print("25%分位数：", df_tech_clean['薪资_均值K'].quantile(0.25), "K")
print("中位数：", df_tech_clean['薪资_均值K'].median(), "K")
print("75%分位数：", df_tech_clean['薪资_均值K'].quantile(0.75), "K")
```

```python
# 看看职位描述长什么样
print(df_tech_clean['职位描述'].head(3).to_string())
```

```python
# 定义我们关心的技能关键词
skill_keywords = [
    'Python', 'SQL', 'MySQL', 'Hive', 'Hadoop', 'Spark', 
    'pandas', 'Excel', 'Tableau', 'PowerBI', 'ETL',
    'Kafka', 'Flink', '数据仓库', '数据清洗', '数据可视化',
    'Linux', 'Git', 'R语言', 'NumPy', 'matplotlib'
]

# 统计每个技能出现的次数
skill_counts = {}
for skill in skill_keywords:
    count = df_tech_clean['职位描述'].str.contains(skill, case=False, na=False).sum()
    skill_counts[skill] = count

# 转成Series排序
skill_series = pd.Series(skill_counts).sort_values(ascending=False)
print(skill_series)
```

```python
# 技能需求频率柱状图
plt.figure(figsize=(12, 6))
plt.bar(skill_series.index, skill_series.values, color='steelblue')
plt.title('数据类岗位技能需求频率')
plt.xlabel('技能')
plt.ylabel('出现次数')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('技能需求.png')
plt.show()
print("图已保存")
```

```python
import jieba
from wordcloud import WordCloud

# 把所有职位描述合并成一段文字
all_text = ' '.join(df_tech_clean['职位描述'].dropna().tolist())

# 用jieba分词
words = jieba.cut(all_text)
word_text = ' '.join(words)

# 停用词列表
stopwords = set(['的', '和', '与', '等', '及', '或', '对', '在', '中', '有', 
                 '为', '以', '并', '能', '是', '不', '了', '也', '都', '将',
                 '负责', '进行', '相关', '具有', '具备', '良好', '优先', '以上',
                 '工作', '能力', '要求', '岗位', '职责', '任职', '参与', '提供',
                 '完成', '管理', '公司', '团队', '业务', '技术', '产品', '系统',
                 '开发', '熟悉', '了解', '掌握', '经验', '年', '者', '如', '可'])

# 生成词云
words_filtered = [w for w in jieba.cut(all_text) if w not in stopwords and len(w) > 1]
word_text_filtered = ' '.join(words_filtered)

wc2 = WordCloud(
    font_path='C:/Windows/Fonts/simhei.ttf',
    width=800,
    height=400,
    background_color='white',
    max_words=80
).generate(word_text_filtered)

plt.figure(figsize=(12, 6))
plt.imshow(wc2, interpolation='bilinear')
plt.axis('off')
plt.title('数据类岗位核心技能词云')
plt.tight_layout()
plt.savefig('词云.png')
plt.show()
print("词云已保存")
```
